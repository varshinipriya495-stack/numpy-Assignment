### A1. Why are NumPy arrays generally preferred over plain Python lists for numerical computation in AI/ML workloads?

NumPy arrays are generally preferred because they are faster and more efficient for numerical calculations. They use less memory than Python lists for storing numerical data. NumPy also provides many mathematical operations and supports vectorized operations, which makes working with large datasets easier and faster.

### A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets?

Vectorization means performing an operation on an entire NumPy array at once instead of using a Python loop for each element. It makes numerical operations faster and more efficient. For large datasets, vectorization reduces execution time and makes the code simpler and easier to work with.

### A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each.

A * B performs element-wise multiplication, where corresponding elements of the two arrays are multiplied. A @ B performs matrix multiplication, where rows of A are multiplied with columns of B. In a neural network, A * B can be used for element-wise operations on values, while A @ B can be used to multiply an input matrix by a weight matrix.

### A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise?

Broadcasting is a NumPy feature that allows arrays with different shapes to be used together in element-wise operations. NumPy compares their shapes from the last dimension to the first. Two dimensions are compatible when they are equal or when one of them is 1. If the dimensions are not compatible, NumPy gives a broadcasting error.

### A5. Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why?

The flatten() function converts an array into a one-dimensional copy, so changes to the new array do not affect the original array. The ravel() function also converts an array into one dimension, but it usually returns a view of the original array when possible. For a memory-constrained application processing large images, I would prefer ravel() because it can avoid creating an extra copy and therefore use less memory.

### A6. For a 2D array, explain axis=0 and axis=1 in an aggregation function such as np.sum() or np.mean(). Illustrate with a small example.

For a 2D array, axis=0 means the operation is performed down the rows, giving a result for each column. Axis=1 means the operation is performed across the columns, giving a result for each row. For example, for [[1, 2], [3, 4]], np.sum(array, axis=0) gives [4, 6], while np.sum(array, axis=1) gives [3, 7].

### A7. Why is it important to normalize or standardize numerical features before training most ML models? Explain normalization vs standardization.

Normalization or standardization helps put numerical features on a similar scale. This can help machine learning models learn more effectively when different features have very different ranges. Normalization usually scales values to a fixed range such as 0 to 1, while standardization transforms values so that they have a mean of 0 and a standard deviation of 1.

### C1. Given a 1D NumPy array of exam scores, use `np.where()` to produce labels 'P' (score >= 40) or 'F' (score < 40). 

````python
import numpy as np

scores = np.array([35, 42, 67, 28, 90, 39, 55])

labels = np.where(scores >= 40, "P", "F")

print("Scores:", scores)
print("Labels:", labels)
````

### C2. Write a function `standardize(x)` that takes a 1D NumPy array and returns its z-score standardized version using only NumPy. 

````python
import numpy as np

def standardize(x):
    return (x - np.mean(x)) / np.std(x)

x = np.array([10, 20, 30, 40, 50])

result = standardize(x)

print("Standardized values:", result)
````

**Explanation:** The function standardizes the values by subtracting the mean and dividing by the standard deviation.

### C3. Given two 2D NumPy arrays representing separate batches with the same number of features, combine them using an appropriate NumPy stacking function and print the resulting shape. 

````python
import numpy as np

batch1 = np.array([[1, 2, 3],
                   [4, 5, 6]])

batch2 = np.array([[7, 8, 9],
                   [10, 11, 12]])

combined = np.vstack((batch1, batch2))

print("Combined shape:", combined.shape)
````

### C4. Given X with shape (100, 5), compute column-wise mean and standard deviation and standardize the matrix in one vectorized expression.

````python
import numpy as np

X = np.random.default_rng(42).normal(size=(100, 5))

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_standardized = (X - mean) / std

print("Mean:", mean)
print("Standard deviation:", std)
print("Standardized shape:", X_standardized.shape)
````

**Explanation:** The columns are standardized using `(X - mean) / std`, where the mean and standard deviation are calculated column-wise with `axis=0`.

### C5. Implement `y_pred = X @ w + b` for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true.

````python
import numpy as np

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])
b = 2.0

y_true = np.array([4, 5, 6])

y_pred = X @ w + b

mse = np.mean((y_true - y_pred) ** 2)

print("Predicted values:", y_pred)
print("MSE:", mse)
````

**Explanation:** The model prediction is calculated using `X @ w + b`, and MSE is calculated as the mean of the squared differences between `y_true` and `y_pred`.

### D1. Data Preprocessing Pipeline 

You are given a raw 1D or 2D NumPy array containing NaN values.

Describe, step by step, the NumPy operations you would apply to:

(a) detect missing values,

(b) handle them appropriately and justify your strategy, and

(c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.

````python
import numpy as np

data = np.array([10.0, 20.0, np.nan, 40.0, 50.0])

# (a) Detect missing values
missing = np.isnan(data)
print("Missing values:", missing)

# (b) Handle missing value
clean_data = np.nan_to_num(data, nan=np.nanmean(data))
print("Clean data:", clean_data)

# (c) Standardize the clean data
mean = np.mean(clean_data)
std = np.std(clean_data)

standardized = (clean_data - mean) / std

print("Standardized data:", standardized)
````

**Explanation:** `np.isnan()` detects the missing value, `np.nanmean()` replaces it with the mean of the available values, and the clean data is standardized using its mean and standard deviation.

### E1. Per-Column Min–Max Normalization 

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

**Explanation:** Per-column min-max normalization scales each feature separately to the range `[0, 1]` using the minimum and maximum values of that column.

````python
import numpy as np

X = np.array([[10, 100],
              [20, 200],
              [30, 300]])

min_values = np.min(X, axis=0)
max_values = np.max(X, axis=0)

normalized = (X - min_values) / (max_values - min_values)

print("Original data:")
print(X)

print("Normalized data:")
print(normalized)
````

**Explanation:** Each column is normalized separately using its own minimum and maximum values, and no explicit loops are used.
